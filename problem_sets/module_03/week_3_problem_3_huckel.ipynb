## Problem 3: What Does an Orbital Look Like? (Nodes)

The `la.eigh()` function doesn't just return energies; it returns **eigenvectors**. These vectors contain the coefficients ($c_1, c_2, c_3...$) that tell us how much each atomic $p$-orbital contributes to the molecular orbital, and whether the wave is pointing "up" (positive) or "down" (negative).

*   If adjacent atoms have the **same sign** (+ and +), the waves add together (constructive interference = bonding).
*   If adjacent atoms have **opposite signs** (+ and -), the waves cancel out. This creates a "Node" (destructive interference = anti-bonding).
